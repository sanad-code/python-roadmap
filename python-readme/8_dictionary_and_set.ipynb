## What are dictionaries and sets? 🤔

- A dictionary is a collection which is unordered, changeable and indexed. In Python dictionaries are written with curly brackets, and they have keys and values.
- A set is a collection which is unordered and unindexed. In Python sets are written with curly brackets.
- Sets are unordered collections of unique elements. Order not guaranteed.
- Dictionaries are unordered collections of key-value pairs.
- Sets and dictionaries are mutable.
- Sets and dictionaries are not hashable, so they can't be used as keys in other dictionaries or sets.
- Sets and dictionaries are not sequences, so they can't be sliced like lists or tuples.


```python
# Simple Example of a Python Dictionary

# Create a dictionary
person = {
    'first_name': 'Omar',
    'last_name': 'Ali',
    'age': 35,
    'country': 'Egypt'
}

# Note the key must be unique and written in quotes
# This will not work: person = {first_name: 'Omar', 'last_name': 'Ali', 'age': 35, 'country': 'Egypt'}

# Print the dictionary
print(person)

# Simple Example of a Python Set

# Create a set
fruits = {'apple', 'banana', 'orange', 'grape', 'banana'}

# Print the set
print(fruits)
```

## How to create a set? 🤔

- A set is created by placing all the items (elements) inside curly braces `{}`, separated by comma, or by using the built-in `set()` function.
- It can have any number of items and they may be of different types (integer, float, tuple, string etc.).
- Sets can contains different data types.
- We can convert a list into a set using the `set()` function.

```python
numbers = { 1, 2,2,2,3,'Ahmed'}
print(numbers)

# Create an empty set
empty_set = set()

# Add elements to the set
empty_set.add(1)
empty_set.add(2)
empty_set.add(2)
empty_set.add(3)
print(empty_set)
```

```python
# Example
"""
Givin a list of numbers, write a Python program to return the number of unique numbers in the list
list = [1, 2, 3, 4, 3, 1, 2, 3, 4, 3, 2]
Expected output: 4
"""
```

```python
list = [1, 2, 3, 4, 3, 1, 2, 3, 4, 3, 2]
unique_numbers = set()
for number in list:
    unique_numbers.add(number)

print(len(unique_numbers))  # Output: 4


# Another simplier way to do it
print(len(set(list)))  # Output: 4
```

## Remove element from a set 🤔

- We can remove elements from a set by using the `remove()` method.
- If the element is not present in the set, then a `KeyError` will be raised.



```python
set1 = {1, 2, 3, 4 ,5 ,5 ,5}
set1.remove(5)
print(set1)

# set1.remove(6)  # This will raise an error because 6 is not in the set
```

## Check if an element is in a set 🤔

- We can check if an element is in a set by using the `in` keyword.
- We can also use the `not in` keyword to check if an element is not in a set.
- We can also use the `issubset()` method to check if a set is a subset of another set.
- We can also use the `issuperset()` method to check if a set is a superset of another set.

```python
set1 = {1, 2, 3, 4 ,5 ,5 ,5}
if 5 in set1:
    print('5 is in the set and it will be removed')
    set1.remove(5)
else:
    print('5 is not in the set')

if 6 in set1:
    print('6 is in the set and it will be removed')
    set1.remove(6)
else:
    print('6 is not in the set')

subset = {2, 3}

if subset.issubset(set1):
    print('subset is a subset of set1')
else:
    print('subset is not a subset of set1')

superset = {1, 2, 3, 4, 5, 6}

if superset.issuperset(set1):
    print('superset is a superset of set1')
else:
    print('superset is not a superset of set1')

```

## Access Dictionary Items 📚

- You can access the items of a dictionary by referring to its key name, inside square brackets.
- There is also a method called get() that will give you the same result.

```python
person = {
    'first_name': 'Omar',
    'last_name': 'Ali',
    'age': 35,
    'country': 'Egypt'
}

# Access the value of the 'first_name' key using the key in square brackets
print(person['first_name'])

# Access the value of the 'first_name' using get()
print(person.get('first_name'))
```

## Dictionary is mutable (changeable) 🔃

- Dictionaries are `mutable`, meaning they can be changed after they are created.
- You can `change`, `add` or `remove` items in a dictionary after it has been created.
- You can change the value of a specific item by referring to its key name.
- You can add an item to a dictionary by `assigning` a value to a new key.
- You can remove an item from a dictionary by using the `pop()`, or `del` keyword.
- You can also remove all the items in a dictionary by using the `clear()` keyword.

```python
person = {
    'first_name': 'Omar',
    'last_name': 'Ali',
    'age': 35,
    'country': 'Egypt'
}

# Change the value of the age key
person['age'] = 36
print(person)

# Add a new key 'email' with the value of 'omar@gmail.com'
person['email'] = 'omar@gmail.com'
print(person)

# Anothor way to add a new key 
person.update({'city': 'Cairo', 'birth_year': 1986})
print(person)

# Remove the key 'city' from the dictionary
person.pop('city')
print(person)

# Remove the key 'birth_year' from the dictionary using del
del person['birth_year']
print(person)

# Remove all the items from the dictionary
person.clear()
print(person)
```

## Iterating through a dictionary 🔄

- You can loop through a dictionary by using a `for` loop.
- When looping through a dictionary, the return value are the keys of the dictionary, but there are methods to return the values as well.
- You can use the `values()`, `keys()`, or `items()` methods to return the values of a dictionary.
- Loop through both keys and values, by using the `items()` method.

```python
person = { 'first_name': 'Omar', 'last_name': 'Ali', 'age': 35, 'country': 'Egypt' }

# Loop through the dictionary keys
for key in person:
    print(key, end = ' ')

print()
# Loop through the dictionary values
for value in person.values():
    print(value, end = ' ')

print()
# Loop through the dictionary keys and values
for key, value in person.items():
    print(key, value, end = ' | ')
```

## Nested Dictionaries 🪺

- A dictionary can also contain many dictionaries, this is called nested dictionaries.

```python
persons = {
    'person1': {
        'first_name': 'Omar',
        'last_name': 'Ali',
        'age': 35,
        'country': 'Egypt'
    },
    'person2': {
        'first_name': 'Ali',
        'last_name': 'Mohamed',
        'age': 25,
        'country': 'Tunisia'
    }
}
print(persons)
```

## In operator with Dictionaries

- The `in` operator can be used to check `if a key is in a dictionary`.
- The `not in` operator can be used to `check if a key is not in a dictionary`.
- The `in` operator can also be used to `check if a value is in a dictionary`.
- The `not in` operator can also be used to `check if a value is not in a dictionary`.

```python
persons = { 'first_name': 'Omar', 'last_name': 'Ali', 'age': 35, 'country': 'Egypt' }

# Check if the key 'first_name' exists in the dictionary
if 'first_name' in persons:
    print('Yes, first_name is one of the keys in the persons dictionary')

# Check if the value 'Ali' exists in the dictionary
if 'Ali' in persons.values():
    print('Yes, Ali is one of the values in the persons dictionary')

# Check if the key 'city' does not exist in the dictionary
if 'city' not in persons:
    print('Yes, city is not one of the keys in the persons dictionary')

# Check if the value 'Tunisia' does not exist in the dictionary
if 'Mohamed' not in persons.values():
    print('Yes, Mohamed is not one of the values in the persons dictionary')
```

## Zip function with Dictionaries 🤐

- The `zip()` function can be used to combine two lists into a dictionary.

```python
list1 = ['first_name', 'last_name', 'age', 'country']
list2 = ['Omar', 'Ali', 35, 'Egypt']

# Create a dictionary from the two lists
persons = dict(zip(list1, list2))
print(persons)
```
